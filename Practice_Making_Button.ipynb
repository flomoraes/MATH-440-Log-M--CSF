```python
import ipywidgets as widgets
from IPython.display import display
import random

# Make compliments
complements = {"Nice job clicking the button",
               "You are so hardworking!",
               "You're a great listener",
               "You have some great problem solving skills",
               "You are a good friend",
               "You are great at persevering",
               "You look great today",
               "You're an amazing researcher"}

# Create button
button = widgets.Button(description = "Click me", button_style = "success")
output = widgets.Output()

# Function controlling click
def clicked(click):
  chosen_compliment = random.choice(tuple(complements))

  with output:
    output.clear_output()
    print(chosen_compliment)


# Link click to function
button.on_click(clicked)

# Display
display(button, output)
```
